```python
#HW10.1. Maximum torque on circular shaft
from sympy import *

d = 67 # mm
sigyield = 233 # MPa
P = 137 # kN

d=d/1000 # m
sigyield=sigyield*10**6 # Pa
P=P*10**3 # N

sigma_x=-P/(pi*d**2/4)
sigma_y=0
T=Symbol('T')
j=(pi*d**4)/32
rho=d/2
tao_xy=T*rho/j

sigma1= (sigma_x+sigma_y)/2 + sqrt(((sigma_x-sigma_y)/2)**2 + tao_xy**2)
sigma2= (sigma_x+sigma_y)/2 - sqrt(((sigma_x-sigma_y)/2)**2 + tao_xy**2)

eq = Eq((sigma1-sigma2)/2, sigyield/2)
sol=solve(eq, T)
print("Maximum torque T on the circular shaft is: ", sol[1]/1000, "kNm")
```

```python
#HW10.2. Failure Criterion Biaxial Loading
from sympy import *

sigma_yield = 60 # MPa

a = Symbol('a')
sigma_x=a
sigma_y=-2*a
tao_xy=0
sigma1= (sigma_x+sigma_y)/2 + sqrt(((sigma_x-sigma_y)/2)**2 + tao_xy**2)
sigma2= (sigma_x+sigma_y)/2 - sqrt(((sigma_x-sigma_y)/2)**2 + tao_xy**2)

eq = Eq((sigma1-sigma2)/2, sigma_yield/2)
sol=solve(eq, a)
print("Tresca criterion: ", sol[1], "MPa")

eq = Eq(sigma1**2+sigma2**2-sigma1*sigma2, sigma_yield**2)
sol=solve(eq, a)
print("von Mises criterion: ", sol[1].evalf(), "MPa")
```

```python
#HW12.1. Buckling I-beam critical load
from sympy import *

E = 284 # MPa
L = 2844 # mm
t = 20 # mm
a = 40 # mm
b = 100 # mm

e=E*1e6 # Pa
l=L/1e3 # m
t=t/1e3 # m
a=a/1e3 # m
b=b/1e3 # m

from math import pi
i1=t*(2*a)**3/12
i2=b*t**3/12+b*t*(a+t/2)**2
i=i1+2*i2

print(i*1e6)


le=l*1
P_cr=pi**2*e*i/(le)**2
print(P_cr/1e3)

le=l*2
P_cr=pi**2*e*i/(le)**2
print(P_cr/1e3)

le=l*0.7
P_cr=pi**2*e*i/(le)**2
print(P_cr/1e3)

le=l*0.5
P_cr=pi**2*e*i/(le)**2
print(P_cr/1e3)
```

```python
#HW12.2. Buckling axial loading find critical load
from sympy import *

L = 1079 # mm
a = 628 # mm
b = 657 # mm
E = 171 # GPa
d = 17 # mm
theta = 40 # degrees

import numpy as np
l=L/1e3 # m
a=a/1e3 # m
b=b/1e3 # m
e=E*1e9 # Pa
d=d/1e3 # m
theta=theta*np.pi/180 # radians

p = Symbol('p')
f_bd=p*l/(a*np.sin(theta))

i=np.pi*(d/2)**4/4    
p_cr=np.pi**2*e*i/(b)**2
sol = solve(f_bd-p_cr,p)  #p_cr=f_bd
print(sol[0]/1e3)
```

```python
#HW12.3. Buckling simple cross-section design
from sympy import *

P = 469 # N
L = 1.173 # m
E = 209 # GPa
FS = 2.2

e=E*1e9

d = Symbol('d')
#i=pi*(d/2)**4/4   #圆
i=d**4/12  #正方形

p_cr=pi**2*e*i/(0.5*L)**2   #l的形式会随连接方式的改变而改变
sol=solve(p_cr-P*FS,d,dict=True)
print(sol[1])
```

```python
#HW12.4. Buckling truss determine largest load
from sympy import *

L = 2221 # mm
d = 10 # mm
E = 219 # GPa
theta = 25 # degrees

l=L*1e-3
d=d*1e-3
e=E*1e9
theta=theta*pi/180

i=pi*(d/2)**4/4
p_cr=pi**2*e*i/(l/2/cos(theta))**2
#f_bc>f_ab
p=symbols('p')
#f_bc=sin(5/4*pi-theta)/sin(2*theta)*2**0.5*p
f_bc=p/2*(1/cos(theta)+1/sin(theta))
sol=solve(p_cr-f_bc,p)
print(sol[0]/1000)
```

```python
#HW12.5. Buckling hollow rectangle critical load
from sympy import *

E = 201 # MPa
L = 3.430 # m
t = 38 # mm
a = 205 # mm
b = 287 # mm

e=E*1e6
l=L
t=t*1e-3
a=a*1e-3
b=b*1e-3

i=a*b**3/12-(a-2*t)*(b-2*t)**3/12

p_cr=pi**2*e*i/l**2
print(p_cr.evalf()/1e3)
```
